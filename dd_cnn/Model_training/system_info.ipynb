```python
import psutil
try:
    import GPUtil
    GPU_AVAILABLE = True
except ImportError:
    GPU_AVAILABLE = False

def get_size(bytes):
    """Convert bytes to GB"""
    return bytes / (1024 ** 3)

print("=" * 50)
print("SYSTEM INFORMATION")
print("=" * 50)
print()

# GPU Information
if GPU_AVAILABLE:
    try:
        gpus = GPUtil.getGPUs()
        if gpus:
            gpu = gpus[0]
            print(f"GPU: {gpu.name} with VRAM: {gpu.memoryTotal:.0f} MB ({gpu.memoryTotal/1024:.1f} GB)")
        else:
            print("GPU: Not detected")
    except:
        print("GPU: Not detected")
else:
    print("GPU: Not detected")

# CPU Information
cpu_count = psutil.cpu_count(logical=True)
cpu_freq = psutil.cpu_freq()
if cpu_freq:
    print(f"CPU: {cpu_count} cores @ {cpu_freq.current:.0f} MHz")
else:
    print(f"CPU: {cpu_count} cores")

# RAM Information
ram = psutil.virtual_memory()
print(f"RAM: {get_size(ram.total):.2f} GB")

# Disk Information
disk = psutil.disk_usage('/')
print(f"DISK: {get_size(disk.total):.2f} GB")

print()
print("=" * 50)
```
